## Set Data Structure

1. set only take unique elements.
2. unordered elements.
3. Mutable.
4. Can contain primitive and non primitive data type.

```python
list=[12,'anik','bolik','sonik',12.5,'anik','12',12]
list2=[1,2,3,4,5]
list01=list+list2
```

```python
setoflist=set(list01)
print(setoflist)
```

```python
setoflist.add(100)
print(setoflist)
```

```python
setoflist.remove(1)
print(setoflist)
```
