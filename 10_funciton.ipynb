## Parameter : a variable that is used in function definition. It is a placeholder for the value that will be passed to the function when it is called.
## Argument : a value that is passed to a function when it is called. It is the actual value that is assigned to the parameter when the function is called.

```python
def add (x,y):
    return x+y;
```

```python
sum=add(5,6);
print(sum)
```

## Function can have multiple return statemnet but only one return statement will be executed. The function will return the value of the first return statement that is executed. The rest of the return statements will be ignored.

**Docstring** : a string that is used to document a function. It is a good practice to write a docstring for a function to explain what the function does, what are the parameters and what is the return value of the function. The docstring should be written in triple quotes and should be the first statement in the function body. The docstring can be accessed using the __doc__ attribute of the function.


```python
def add(a, b):
    """this is a docstring. It is used to explain what the function does."""
    return print(f"Sum is = {a+b}")
```

```python
add(5,6)
```

**Global and Local variable** : a variable that is defined outside the function is called global variable and a variable that is defined inside the function is called local variable. A local variable can only be accessed inside the function and a global variable can be accessed anywhere in the code. If a local variable has the same name as a global variable, then the local variable will be used inside the function and the global variable will be used outside the function.

## Built in Function : a function that is already defined in Python and can be used without importing any module. For example, print(), len(), type() etc. are built in functions.

## Lambda Function
1. A lambda function is a small, **anonymous** function that can take any number of arguments but can only have one expression.
2. Lambda functions are used when you need a simple function for a short period of time and don't want to formally define it using the `def` keyword.
3. The syntax for a lambda function is: `lambda arguments: expression`
4. Example: `x = lambda a, b : a * b`

```python
sum1=lambda a,b:a+b;
print(sum1(5,6))
```

## map(), filter(),sorted()

```python
my_list=[1,2,3,4,5]
squared_list=list(map(lambda x:x*x,my_list))
print(squared_list)
```

```python
odd_number=list(filter(lambda x:x%2!=0,my_list))
print(odd_number)
```

```python
## sorted()
names=["anik","rofiq","sajib","anik khondokar","yamin-hawlader","sofiq"]
sorted_names=sorted(names)
print(sorted_names)
```

```python
# lengthwise sorted
sorted_names2=sorted(names,key=lambda i:len(i))
print(sorted_names2)
```
