## Dictionary Data Structure (Python)

## 🔹 1. Definition

A **dictionary** is a collection of **key-value pairs** where each key is unique. It allows fast retrieval of values based on their corresponding keys.

## 🔹 2. Characteristics

* **Unordered**: Items do not have a fixed order.
* **Mutable**: You can modify the dictionary after creation.
* **Dynamic**: The size can grow or shrink as needed.

## 🔹 3. Syntax

In Python, dictionaries are defined using curly braces `{}`:

```python
my_dict = {
    'key1': 'value1',
    'key2': 'value2'
}
```

Dictionaries are a powerful and flexible data structure in Python that provide **fast access, easy modification, and efficient storage** using unique keys.


````python
#Example
students={
    'name':"Anik Khondokar",
    'name':"Sara Ahmed",
    'age': 25,
    'city': "Dhaka"
}
print(students)
````

````python
print(students['name'])
````

````python
if 'name' in students:
    print(students['name'])
````

````python
#for all the values
print(students.values())
````

````python
#for all the keys
print(students.keys())
````

````python
#Accessing value using keys
print(students['name'])
````

````python
#What is the key does not exist
print(students['country'])
````

````python
## .get() method to avoid error
print(students.get('country'))
````

````python
# add a new key-value pair
students['country']="Bangladesh"
print(students)
````

````python
#delete a key-value pair
del students['age']
print(students)
````

### Adding or deleting key-value pairs in a dictionary is a inplace operation meaning it modifies the original dictionary without creating a new one.

## Iterating over a dictionary

````python
students.items()
````

````python
# so its actually a list of tuples where each tuple contains a key-value pair from the dictionary.
# This allows us to iterate over the dictionary and access both keys and values at the same time. 
# We are using tuple unpacking. 
for key,value in students.items():
    print(f"Key: {key}, Value: {value}")
````

## Creating a dictionary using for loop

````python
subjects=['Math','Physics','Chemistry']
scores=[85,90,95]

my_dict={}

for i in range(len(subjects)):
    my_dict[subjects[i]]=scores[i]

print(my_dict)
````

## Creating a dictionary with a dictionary comprehension

````python
# what is zip() function
# zip() function takes iterables (can be zero or more), aggregates them in a tuple, and returns it. 
# The zip() function is used to combine two or more iterables (like lists, tuples, etc.) element-wise into a single iterable of tuples. 
# Each tuple contains one element from each of the input iterables, grouped together based on their position.                   

for k,v in zip(subjects,scores):
    print(f"Subject: {k}, Score: {v}")
````

````python
my_dict2={k: v for (k, v) in zip(subjects, scores)};
print(my_dict2)
````

## Dictionary of list


````python
favorite_foods={
    'anik': ['Pizza', 'Burger', 'Pasta'],
    'sara': ['Sushi', 'Ramen', 'Tempura'],
    'john': ['Steak', 'Fries', 'Salad']
}
````

````python
for key,value in favorite_foods.items():
    print(
        f"\n{key.capitalize()}'s favorite foods are"
    )
    for food in value:
        print(f'\n- {food}')
````

## Dictionary of Dictionarys

````python
alumni={
    'anik': {
        'graduation_year': 2020,
        'degree': 'Computer Science',
        'university': 'Jagannath University'
    },
    'sara': {
        'graduation_year': 2019,
        'degree': 'Business Administration',
        'university': 'University of Dhaka'
    },
    'john': {
        'graduation_year': 2021,
        'degree': 'Mechanical Engineering',
        'university': 'BUET'
    }
}
````

````python
for key,value in alumni.items():
    print(
        f"{key.title()}',{value['degree']}\n{value['university']}\n"
    )
````

````python
# adding new key-value pair to the dictionary
alumni['maria'] = {
    'graduation_year': 2022,
    'degree': 'Electrical Engineering',
    'university': 'NSU'
}
````

````python
print(alumni)
````

````python
students.update(alumni)
print(students)
````
