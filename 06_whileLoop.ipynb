## While loop :

```python
#printing 1 to 10 using while loop
i=1
while i<=10 :
    print(i,end=" ")
    i=i+1
```

## Short Hand

```python
i=1
while i<=10 :
    print(i,end=", ")
    i+=1
```

```python
# sum 1 to 100 using while loop
sum=0
i=1
while i<=100:
    sum=sum+i
    i+=1
print(f"sum = {sum}")
```

## break, continue and pass statements
1. These are called loop control statements.
2. **break :** Stops executing the remaining steps.
3. **continue:** return control to the beginning of the loop.
4. **pass:** do nothing. **just a placeholder**

```python
#break statements in for loop
for i in range(1, 15):
    print(i)
```

```python
#break statements in for loop
for i in range(1, 15):
    print(i)
    if i==5 :
        break
```

```python
#continue statements in for loop
for i in range(1, 10):
    if i==5:
        continue
    print(i)
```

```python
# pass statements
for i in range (1 , 11):
    if i==5:
        pass
    print(i)
```

## Problem solving

```python
# stop when user enter some strings
while True:
    s=input("Enter a string: ")
    if s!="":
        break
print("User input is "+ " "+s)
```

## Function in Python
It is not like a blackbox. We actually know what is going on inside a function.

```python
def getSum():
    sum=0
    for i in range(1,11):
        sum+=i
    print(sum)
    
getSum();
```

## with parameter

```python
def getSum(num):
    sum=0
    for i in range(num):
        sum+=i
    return sum
    
sum=getSum(100);
print(sum)
```
