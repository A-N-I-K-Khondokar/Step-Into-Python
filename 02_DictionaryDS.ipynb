## Dictionary Data Structure

1. Can contain heterogeneous primitive and non-primitive data structure.
2. Can store multiple list within one DS.
3. One single object for multiple list.
4. Key:value pair.
5. `empty-dict={}`
6. Elements are unordered and can not be extracted by integer position.
7. Have mutability. So dictionary DS is mutable.

```python
dict={
    "key1": ["value1", "value2"],
    "key2": [1,'value3'],
    "key3": [1,2,3]
}
```

```python
dict.keys()
```

```python
dict["key1"]
```

```python
dict['key2']
```

```python
dict.update({"key4": "value4"})
```

```python
dict.values()
```
