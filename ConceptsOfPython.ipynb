## Class : Each object belong to a class of datatype. So, all the data type like `int, float, str, bool` they belong to a corresponding class. 
    - 2 and 45 belongs to `int` class.
    - -34.45 and 56.34 is belongs to `float` class.
    - True and False belongs to `bool` class.
    - “This one is string” belong to `str` class.
