## Tuple Data Structure in Python
1. **Definition**: A tuple is an ordered, **immutable** collection of items. It can contain elements of different data types.

```python
# Accesssing Tuple Elements
a=(1,'r',3.5,True)
a[1]
```

```python
for i in a :
    print(i)
```

## Tuple unpacking
**Definition**: Tuple unpacking is a **convenient** way to assign the elements of a tuple to multiple variables in a single statement.

```python
students = ('Alice', 'Bob', 'Charlie')
name1, name2, name3 = students
print(name1)  # Output: Alice
print(name2)  # Output: Bob
print(name3)  # Output: Charlie
```

```python
## what happen if i assign more variable or less variable for unpacking
name1,name2,name3,name4=students
```

```python
name1,name2=students
```

## Extended unpacking
**Definition**: Extended unpacking allows you to unpack a tuple into variables while also capturing any remaining elements into a list using the `*` operator.

```python
a,*b,c=(1,2,3,4,5)
print(*b)
b
```
