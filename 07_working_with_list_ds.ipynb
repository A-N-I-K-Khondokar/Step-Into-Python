## strip() method of string
1. strip() method removes any leading and trailing characters (space is the default leading character to remove)
2. strip() method does not change the original string, it returns a new string with the specified characters removed.
3. strip() method can take an optional argument which is a string specifying the set of characters to be removed. If no argument is provided, it will remove whitespace characters by default. 


```python
fruties=['apple   ','banana   ','    orange','grape','kiwi']
fruties
```

```python
fruties=[i.strip() for i in fruties]
fruties
```

```python
fruties=[i.capitalize() for i in fruties]
fruties
```

## slicing opetator without the strat and end index
1. when we use slicing operator without the start and end index, it creates a copy of the original list. 
2. it is a common way to create a copy of a list in python. 
3. **it is also known as shallow copy of a list.**
4. it is important to note that when we create a copy of a list using slicing operator, it creates a new list object in memory, but the elements of the list are still references to the same objects in memory. 
5. this means that if we modify an element in the original list, **it will also modify the corresponding element in the copied list**, and vice versa. 
6. if we want to create a deep copy of a list, we can use the **copy** module in python, which provides a deepcopy() function that creates a new list object and recursively copies all the elements of the original list, including any nested lists or objects.

```python
cars=['bmw','audi','mercedes','toyota','honda']
cars2=cars[:]
print(cars2)
```
