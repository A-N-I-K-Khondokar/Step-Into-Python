## List Data Structure

1. It is a non-primitive data structure. In list DS, we are wrapping a primitive DS.
2. Collection of primitive DS.
3. Create a empty list.
4. Collection can contain heterogenous primitive / non-primitive DS.
5. Can be modified.

```python
list1=[1, 2, 3, 4, 5]
```

```python
print(list1)
```

```python
list2=('Rofiq', 'Anik', 'Khondokar')
print(list2)
```

```python
print(list2[2])
```
