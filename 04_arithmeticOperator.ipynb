## AND operation
| A | B | A and B |
| --- | --- | --- |
| 0 | 0 | T |
| 0 | 1 | F |
| 1 | 0 | F |
| 1 | 1 | F |

## OR operation
| A | B | A or B |
| --- | --- | --- |
| 0 | 0 | F |
| 0 | 1 | T |
| 1 | 0 | T |
| 1 | 1 | T |
